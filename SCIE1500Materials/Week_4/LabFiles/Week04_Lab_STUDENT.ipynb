## Part A: Derivatives as Rates of Change (~20 min)

The derivative $Y'(T) = \frac{dY}{dT}$ tells us how yield changes per degree of temperature change.

Applying the power rule to $Y(T) = -0.05T^2 + 8T - 200$:
$$Y'(T) = -0.1T + 8$$

This is the **marginal yield** — the extra grams per batch gained by raising temperature 1°C.

````python
# A.1 — Define yield function and derivative
def Y(T):
    'Drug yield (grams/batch).'
    return -0.05 * T**2 + 8 * T - 200

def dY(T):
    'Marginal yield (grams per batch per °C).'
    return -0.1 * T + 8

# Evaluate at three temperatures
print(f"{'T (°C)':>8} | {'Y(T) (g)':>10} | {'Y\'(T) (g/°C)':>14} | Yield trend")
print("-" * 55)
for T in [60, 80, 100]:
    trend = "increasing" if dY(T) > 0 else "decreasing"
    print(f"{T:>8} | {Y(T):>10.1f} | {dY(T):>14.2f} | {trend}")
````

## Part B: Finding the Optimal Temperature (~25 min)

A **critical point** occurs where $Y'(T) = 0$ — a candidate for maximum or minimum yield.

Solving $-0.1T + 8 = 0$ gives $T_{\text{opt}} = 80$ °C.

The **second derivative** $Y''(T)$ confirms the type:
- $Y'' < 0$: concave down → **maximum** ✓
- $Y'' > 0$: concave up → minimum

````python
# B.1 — Critical point and second derivative test
def d2Y(T):
    'Second derivative (curvature).'
    return -0.1      # constant — negative everywhere → always concave down

T_opt = 8 / 0.1    # solve -0.1T + 8 = 0  →  T = 80
Y_opt = Y(T_opt)
Y2    = d2Y(T_opt)

print(f"Critical point: T_opt = {T_opt:.1f} °C")
print(f"Y(80) = {Y_opt:.1f} g/batch  (maximum yield)")
print(f"Y''(80) = {Y2:.1f}  (< 0 → confirmed maximum)")
print(f"\nFeasible? 50 ≤ {T_opt:.0f} ≤ 110 → {50 <= T_opt <= 110}")
````

Plot yield curve with optimal point and safe range.

````python
# B.2 — Plot yield curve with optimal point and safe range
fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(T_vals, Y(T_vals), "steelblue", lw=2, label="Yield Y(T)")
ax.plot(T_opt, Y_opt, "ro", ms=12, label=f"Optimum: T={T_opt:.0f}°C, Y={Y_opt:.1f} g")
ax.axvline(50,    color="gray", ls="--", alpha=0.5, label="Safe range [50, 110]°C")
ax.axvline(110,   color="gray", ls="--", alpha=0.5)
ax.set_xlabel("Temperature T (°C)")
ax.set_ylabel("Yield (g/batch)")
ax.set_title("Drug Yield: Optimal Temperature")
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
````

## Part D: Symbolic Differentiation with SymPy (~10 min)

Throughout this lab you found $Y'(T)$ **by hand**, applying the power rule term by term. Python has a library called **SymPy** that can do this kind of algebra automatically — not just evaluate a formula at numbers (which is all `numpy` ever does), but manipulate the *symbols themselves*, the same way you do on paper. This is called **symbolic computation**, and it's the standard tool you'll reach for whenever you need to differentiate or integrate in Python for the rest of this unit.

Run the cell below to have SymPy differentiate $Y(T)$ symbolically and confirm it matches your by-hand answer from Parts A and B.

````python
# D.1 — Symbolic differentiation with SymPy
import sympy as sp

T_sym = sp.symbols('T')
Y_expr = -0.05*T_sym**2 + 8*T_sym - 200

dY_expr  = sp.diff(Y_expr, T_sym)        # first derivative, symbolic
d2Y_expr = sp.diff(Y_expr, T_sym, 2)     # second derivative, symbolic

print(f"Y(T)   = {Y_expr}")
print(f"Y'(T)  = {dY_expr}   ← same as the power-rule result from Part A")
print(f"Y''(T) = {d2Y_expr}  ← same constant as d2Y() in Part B")

# Solve Y'(T) = 0 symbolically — same critical point as Part B
critical_points = sp.solve(sp.Eq(dY_expr, 0), T_sym)
print(f"\nCritical point(s): T = {critical_points}")
````

**Why this matters:** `sp.diff()` performs **symbolic differentiation** — it works with the algebraic expression itself (the power rule, product rule, chain rule, etc. — the exact same rules you use on paper), rather than approximating a derivative from numbers. This is fundamentally different from `numpy`, which only ever computes with actual numbers, never symbols. SymPy is a **symbolic computation** library more broadly: it can also `solve()` equations exactly, simplify algebra, and — as you'll see in Weeks 6 and 7 — `integrate()` expressions symbolically too.
