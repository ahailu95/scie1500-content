## 📋 Scenario

![Avocado market comparison: domestic only vs with exports](../images/W7B_avocado_market.svg)

The Australian avocado market:

- **Demand:** $Q_d = 600 - 4P$ (million avocados/yr)
- **Supply:** $Q_s = -120 + 2P$ (million avocados/yr)

An export deal is struck: new effective demand becomes $Q_d^{\text{export}} = 720 - 4P$.

```python
# Run first — loads libraries for this session
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams.update({'figure.dpi': 120, 'font.size': 11})

# Avocado domestic market
def demand_P(Q):
    'Domestic inverse demand.'
    return (600 - Q) / 4

def supply_P(Q):
    'Inverse supply.'
    return (Q + 120) / 2

def demand_export_P(Q):
    'Export inverse demand: Qd_export = 720 - 4P.'
    return (720 - Q) / 4
```

## Part A: Domestic Market Baseline

Before the export deal, domestic buyers and Australian growers determine the market outcome. Find the equilibrium by setting $Q_d = Q_s$, then solve for $P^*$ and $Q^*$ before running the check below.

> **Presentation prep:** Show your algebra, then explain what the equilibrium price and quantity mean for the Australian avocado market. Keep units explicit: quantities are in **millions of avocados per year** and prices are **dollars per avocado**.

```python
# Domestic equilibrium: 600 - 4P = -120 + 2P → 6P = 720 → P*=120, Q*=120
P_dom = 120
Q_dom = 120
print(f"Domestic equilibrium: P*=${P_dom}, Q*={Q_dom}M avocados")
```

## Consumer and Producer Surplus in the Domestic Market

Consumer surplus (CS) is the area below inverse demand and above the market price. Producer surplus (PS) is the area above inverse supply and below the market price. With straight-line curves, both are triangles.

> **Presentation prep:** Identify the base and height of each triangle before calculating. Explain whose willingness to pay is represented by CS and whose gains from selling are represented by PS.

```python
# Domestic CS and PS
CS_dom = 0.5 * Q_dom * (demand_P(0) - P_dom)
PS_dom = 0.5 * Q_dom * (P_dom - supply_P(0))
print(f"Domestic CS: ${CS_dom:,.1f}M")
print(f"Domestic PS: ${PS_dom:,.1f}M")
print(f"Total Surplus: ${CS_dom + PS_dom:,.1f}M")
```

## Part B: Export Demand Shock

The export deal raises effective demand from $Q_d = 600 - 4P$ to $Q_d^{\mathrm{export}} = 720 - 4P$. This is a rightward shift of the demand curve: at every price, domestic and export buyers together want 120 million more avocados than before.

Find the new equilibrium by setting the expanded demand equal to supply. Then explain why the price, quantity, CS, and PS change.

> **Presentation prep:** Distinguish between the **aggregate CS** calculated from the expanded demand curve and the surplus received by existing domestic buyers. The two are not the same.

```python
# Export equilibrium: 720 - 4P = -120 + 2P → 6P = 840 → P*=140, Q*=160
P_exp = 140
Q_exp = 160
print(f"Export equilibrium: P*=${P_exp}, Q*={Q_exp}M avocados")

CS_exp = 0.5 * (720 / 4 - P_exp) * Q_exp   # demand_export_P(0) = 180
PS_exp = 0.5 * Q_exp * (P_exp - supply_P(0))
print(f"\nExport CS: ${CS_exp:,.1f}M  (change: ${CS_exp - CS_dom:+,.1f}M)")
print(f"Export PS: ${PS_exp:,.1f}M  (change: ${PS_exp - PS_dom:+,.1f}M)")
print(f"Net welfare change: ${(CS_exp + PS_exp) - (CS_dom + PS_dom):+,.1f}M")
```

## Part C: Read the Welfare Comparison

The two panels below compare the original domestic market with the market after export demand is added.

> **Presentation prep:** Use the figure to explain the rightward demand shift, the movement from $(Q,P)=(120,120)$ to $(160,140)$, and why an increase in aggregate CS does not by itself show that existing domestic consumers are better off.

The chart below shows the domestic vs export equilibrium.

```python
# Plot: domestic vs export equilibrium
Q_vals = np.linspace(0, 250, 300)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))

for ax, (P_star, Q_star, Pinv_d, label, cs, ps) in [
    (ax1, (P_dom, Q_dom, demand_P,        "Domestic",  CS_dom, PS_dom)),
    (ax2, (P_exp, Q_exp, demand_export_P, "With Export", CS_exp, PS_exp))]:
    Q_sh = np.linspace(0, Q_star, 300)
    ax.plot(Q_vals, Pinv_d(Q_vals), "seagreen", lw=2.5, label="Demand")
    ax.plot(Q_vals, supply_P(Q_vals), "coral", lw=2.5, label="Supply")
    ax.fill_between(Q_sh, Pinv_d(Q_sh), P_star, alpha=0.25, color="seagreen", label=f"CS=${cs:.0f}M")
    ax.fill_between(Q_sh, P_star, supply_P(Q_sh), alpha=0.25, color="coral", label=f"PS=${ps:.0f}M")

    # Triangle centroids keep the labels inside their corresponding shaded areas.
    label_x = Q_star / 3
    cs_label_y = (Pinv_d(0) + 2 * P_star) / 3
    ps_label_y = (supply_P(0) + 2 * P_star) / 3
    ax.text(label_x, cs_label_y, "Consumer\nSurplus", color="seagreen", ha="center", va="center", fontweight="bold", fontsize=9)
    ax.text(label_x, ps_label_y, "Producer\nSurplus", color="firebrick", ha="center", va="center", fontweight="bold", fontsize=9)

    ax.plot(Q_star, P_star, "ko", ms=9, zorder=5, label=f"Eq. (Q={Q_star:.0f}M, P=${P_star:.0f})")
    ax.axhline(P_star, color="black", ls="--", alpha=0.35)
    ax.axvline(Q_star, color="black", ls="--", alpha=0.35)
    ax.set_title(f"Avocado Market - {label}")
    ax.set_xlabel("Quantity (M avocados/yr)"); ax.set_ylabel("Price ($/avocado)")
    ax.set_ylim(0, 200); ax.legend(fontsize=8); ax.grid(alpha=0.3)

plt.tight_layout(); plt.show()
```
